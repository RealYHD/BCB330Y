```python
import csv
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from collections import defaultdict
import pandas as pd
from IPython.display import JSON
from matplotlib import colors
```

## Data Input

```python
chromosome_rohs = {
    "X": defaultdict(list)
}
for i in range(1, 24):
    chromosome_rohs[str(i)] = defaultdict(list)

with open(os.path.abspath("ROH Data/1500kb - Minimum 100 SNPs.csv"), "r") as file_handle:
    header = None
    reader = csv.reader(file_handle)
    for row in reader:
        if not header:
            header = [f"{i}. {name}" for i, name in enumerate(row)] # Add numerical indices to account for repeating headers
            continue
        for index, item in enumerate(row):
            try:
                item = float(item)
            except ValueError:
                item = str(item)
            chromosome_rohs[row[int(header.index("1. Chromosome Name"))]][header[index]].append(item)
for rohs in chromosome_rohs.values():
    rohs = dict(rohs)
    display(pd.DataFrame(rohs))
```

```python
start_pos_key = "2. Start Position"

for chrom, rohs in chromosome_rohs.items():
    start_positions = np.array(sorted(rohs[start_pos_key]))
    start_positions = start_positions.reshape(-1, 1)

    n_components = 4
    if len(start_positions) < n_components:
        continue
    
    gmm = GaussianMixture(n_components=n_components)
    gmm.fit(start_positions)
    fig, axs = plt.subplots(1, 1, figsize=(15,5), tight_layout=True)
    
    counts, bins, patches = axs.hist(start_positions, bins=64)
    bin_norm = colors.Normalize(0, n_components)
    for count, bin, patch in zip(counts, bins, patches):
        color = plt.cm.viridis(bin_norm(gmm.predict(np.array([[bin]]))))
        patch.set_facecolor(color)
    plt.savefig(f"chr{chrom}_histogram.svg", format="svg")
    plt.close()

    predictions = gmm.predict(start_positions)
    with open(f"roh_start_gmm_chr{chrom}.csv", "w") as file_handle:
        writer = csv.writer(file_handle)
        writer.writerow([*header, "GMM Group"])
        for index, start_pos in enumerate(rohs[start_pos_key]):
            row = []
            for col in header:
                row.append(rohs[col][index])
            row.append(predictions[index])
            writer.writerow(row)
```
